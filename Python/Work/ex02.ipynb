```python
# 실습 1: 학생 점수 평균 구하기
def average(scores):
    return sum(scores) / len(scores) # Python에서는 avg 함수 존재 X
# sum(): 합계를 구하는 함수
# len(): 길이를 구하는 함수

scores = [90, 80, 100] # List
print(average(scores))
```

```python
# 실습 2: BMI 계산기
def bmi(height , weight):
    h = height / 100 # cm → m 변환
    return weight / (h * h)  

print(bmi(170, 70))
```

```python
# 실습 3: 두 수 중에서 큰 수를 반환
def max_number(a, b):
    if a > b:
        return a
    return b

print(max_number(30, 50))

print(type(max_number))
print(id(max_number))

fun = max_number
fun(165, 60)
```

```python
import csv

total = 0
count = 0

with open("student.csv", "r", encoding = "utf-8") as file:
    reader = csv.reader(file)
    next(reader) # 첫 번째 행(헤더) 건너뛰기

    for row in reader:
        total += int(row[1]) # 국어 점수
        count += 1
        # print(row)

avg = total / count
print("국어 평균 :", avg)
```
