### 조건 검색(Boolean Indexing)

````python
import numpy as np

scores = np.array([70, 80, 90, 100, 60])
````

````python
print(scores >= 80) # boolean indexing: 80점 이상이면 True, 아니면 False

print(scores[scores >= 80]) # 80점 이상인 점수만 가져오기
````

### 다양한 조건

````python
print(scores[(scores >= 60) & (scores <= 90)])
````

> 주의: NumPy에서는 `and`, `or` 대신 `&`, `|`를 사용한다.

#### where()
> 조건에 맞는 위치(Index)를 반환하거나 조건에 따라 값을 변경할 때 사용하는 함수

#### 위치 찾기

````python
print(scores)

print(np.where(scores <= 70)) # index 반환
````

````python
result = np.where(scores >= 80, "합격", "불합격")

print(result)
````

````python
# 80점 이상만 출력하기
import numpy as np

scores = np.array([65, 75, 85, 95, 55])
print(scores[scores>=80])
````

````python
# 70점 이하만 출력하기
print(scores[scores <= 70])
````

````python
# 80점 이상이면 "PASS", 아니면 "FAIL" 출력하기
result = np.where(scores >= 80, "PASS", "FAIL")

print(result)
````

### 표준 편차(Standard Deviation)
> **데이터가 평균으로부터 얼마나 떨어져 있는지를 나타내는 값이다.**
> 

쉽게 말하면

> **데이터가 얼마나 고르게 분포되어 있는지를 나타내는 지표이다.**

#### 예제 1
학생들의 국어 점수이다.

```
80
90
100
```

평균은 

```
90점
```

각 점수가 평균에서 얼마나 떨어져 있는지 보면

```
80 → 평균보다 10점 작음

90 → 평균과 같음

100 → 평균보다 10점 큼
```

↓

점수 차이가 크지 않다.

↓

표준편차도 작다.

### 평균과 함께 보기

````python
import numpy as np

arr = np.array([80, 90, 100])

print("평균 :", arr.mean())
print("표준편차 :", arr.std())
````

### 난수 생성

#### randint()
> **지정한 범위의 정수(Integer)를 랜덤하게 생성하는 함수이다.**
> 

※ **같은 숫자가 나올 수도 있다.**

````python
np.random.randint(1, 46, 6) # 1~45까지 6개의 난수 생성
````

#### np.random.choice()
> 지정한 데이터(배열)에서 랜덤하게 값을 선택하는 함수

````python
# 예제1
import numpy as np

fruits = ["사과", "바나나", "포도"]

print(np.random.choice(fruits))
````

````python
# 예제2
# replace=False 옵션을 주면 같은 숫자가 나오지 않는다.
import numpy as np

lotto = np.random.choice(range(1, 46), 6, replace = False)

print(lotto)
````
